```python
import polars as pl
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

ROOT = Path("")   # racine du projet ; mets Path(".") si le notebook est à la racine
PARTS = str(ROOT / "data/interim/parts/*.parquet")

pl.Config.set_tbl_rows(40)
pl.Config.set_fmt_str_lengths(70)
pl.Config.set_tbl_width_chars(180)

MASK = r"X{2,}"
CAT_COLS = ["product", "sub_product", "issue", "sub_issue", "company", "state", "tags",
            "submitted_via", "company_response_to_consumer", "timely_response",
            "company_public_response"]

# ---------- Base : plaintes avec récit ----------
base = (
    pl.scan_parquet(PARTS)
    .with_columns(
        pl.col("consumer_complaint_narrative").str.strip_chars().alias("narr"),
        pl.col("date_received").str.to_date("%Y-%m-%d").alias("date"),
    )
    .filter(pl.col("narr").str.len_chars() > 0)
    .with_columns(pl.col("date").dt.year().alias("year"))
)

# ---------- Table de stats (sans le texte) ----------
stats = (
    base.with_columns(
        pl.col("narr").str.len_chars().alias("n_chars"),
        pl.col("narr").str.count_matches(MASK).alias("n_masks"),
        pl.col("narr").str.replace_all(MASK, " ").str.count_matches(r"[A-Za-z]{2,}").alias("n_words"),
    )
    .select(["complaint_id", "year", "n_chars", "n_masks", "n_words", *CAT_COLS])
    .with_columns([pl.col(c).cast(pl.Categorical) for c in CAT_COLS])
    .collect(engine="streaming")
)
N = stats.height
print(f"{N:,} récits | {stats.estimated_size('mb'):.0f} Mo en RAM")

# ---------- 1. Value counts de chaque colonne catégorielle ----------
def vc(col, top=25):
    t = (
        stats.group_by(col).len().sort("len", descending=True)
        .with_columns(
            (pl.col("len") / N * 100).round(2).alias("pct"),
            (pl.col("len").cum_sum() / N * 100).round(1).alias("pct_cumul"),
        )
    )
    print(f"\n### {col} — {t.height} valeurs distinctes | nulls : {stats[col].null_count():,}")
    display(t.head(top))

for c in CAT_COLS:
    vc(c)

# ---------- 2. Longueurs ----------
print("\n### Longueurs (n_words = mots réels, hors XXXX)")
display(stats.select("n_chars", "n_words", "n_masks")
             .describe(percentiles=(0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99)))

print("\n### Nombre de récits sous chaque seuil de mots réels")
display(pl.DataFrame({
    "seuil_mots": [3, 5, 10, 20, 30, 50],
    "n_sous_seuil": [int((stats["n_words"] < k).sum()) for k in [3, 5, 10, 20, 30, 50]],
}).with_columns((pl.col("n_sous_seuil") / N * 100).round(2).alias("pct")))

# ---------- 3. Produit x année en % ----------
print("\n### Part de chaque produit par année (%)")
share = (
    stats.group_by("year", "product").len()
    .with_columns((pl.col("len") / pl.col("len").sum().over("year") * 100).round(1).alias("pct"))
    .pivot(on="year", index="product", values="pct", sort_columns=True)
    .fill_null(0).sort("product")
)
display(share)

# ---------- 4. Graphiques ----------
fig, ax = plt.subplots(1, 2, figsize=(15, 4))
yc = stats.group_by("year").len().sort("year")
ax[0].bar(yc["year"], yc["len"])
ax[0].set_title("Récits par année")
ax[0].set_yscale("log")

cap = stats["n_words"].quantile(0.99)
ax[1].hist(stats["n_words"].clip(upper_bound=cap), bins=100)
ax[1].set_title("Distribution des mots réels par récit (coupée au p99)")
ax[1].set_xlabel("mots")
plt.tight_layout()
plt.show()

pdf = share.to_pandas().set_index("product").T
pdf.plot(kind="bar", stacked=True, figsize=(15, 6), colormap="tab20")
plt.title("Répartition des produits par année (%)")
plt.legend(bbox_to_anchor=(1.01, 1), loc="upper left", fontsize=8)
plt.tight_layout()
plt.show()
```

```python
def show_short(lo, hi, n=15, seed=0):
    ids = stats.filter(pl.col("n_words").is_between(lo, hi)).sample(n, seed=seed)["complaint_id"].to_list()
    ex = base.filter(pl.col("complaint_id").is_in(ids)).select("complaint_id", "product", "narr").collect()
    print(f"\n=== {lo} à {hi} mots réels ===")
    for r in ex.iter_rows(named=True):
        print(f"[{r['product']}] {r['narr']!r}")

show_short(0, 3)
show_short(4, 10)
show_short(11, 20)
```
